## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

- **Percepción (S):** tiempo de espera del ticket, nivel de urgencia y condición de cliente premium.
- **Acciones (A):** mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3.
- **Entorno (E):** mesa de ayuda, clientes, técnicos de soporte, sistema de tickets y servicios tecnológicos.
- **Objetivo:** asignar cada ticket al nivel de soporte adecuado para resolverlo de manera rápida y eficiente.
- **Medida de desempeño (P):** tiempo promedio de atención, cantidad de tickets resueltos, satisfacción del cliente y cumplimiento de los tiempos establecidos.

### Justificación de las reglas

Las reglas priorizan los tickets de urgencia alta porque pueden afectar seriamente el servicio del cliente. Los casos de urgencia media se evalúan considerando el tiempo de espera y la condición premium. Los tickets de urgencia baja se mantienen en el primer nivel, excepto cuando la espera es elevada, para evitar retrasos excesivos y distribuir mejor el trabajo del equipo.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    nivel_urgencia = nivel_urgencia.lower().strip()

    if tiempo_espera_minutos < 0:
        return "revisar ticket", "el tiempo de espera no puede ser negativo"

    if nivel_urgencia not in ["baja", "media", "alta"]:
        return "revisar ticket", "el nivel de urgencia ingresado no es válido"

    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "el ticket presenta urgencia alta"

    if nivel_urgencia == "media":
        if cliente_premium or tiempo_espera_minutos >= 30:
            return "escalar a nivel 2", "urgencia media con prioridad premium o espera de 30 minutos"
        else:
            return "mantener en nivel 1", "urgencia media con tiempo de espera menor a 30 minutos"

    if cliente_premium and tiempo_espera_minutos >= 45:
        return "escalar a nivel 2", "cliente premium con tiempo de espera elevado"

    if tiempo_espera_minutos >= 60:
        return "escalar a nivel 2", "el ticket de urgencia baja superó 60 minutos de espera"

    return "mantener en nivel 1", "urgencia baja y tiempo de espera dentro del límite"
```

### Simulación y pruebas

Se prueban seis tickets con diferentes niveles de urgencia, tiempos de espera y tipos de cliente. Se incluyen casos límite y situaciones donde las condiciones no coinciden, como un ticket de urgencia baja perteneciente a un cliente premium con una espera elevada.

```python
casos_prueba = [
    (5, "alta", False),
    (10, "media", True),
    (30, "media", False),  # Caso límite para urgencia media
    (20, "baja", False),
    (50, "baja", True),    # Baja urgencia, premium y espera elevada
    (60, "baja", False)    # Caso límite para urgencia baja
]

for numero, (tiempo, urgencia, premium) in enumerate(casos_prueba, start=1):
    accion, motivo = agente_soporte(tiempo, urgencia, premium)

    print(
        f"Prueba {numero}: espera={tiempo} min | "
        f"urgencia={urgencia} | premium={premium} | "
        f"decisión={accion} | motivo={motivo}"
    )
```
