## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

- **Percepción (S):** tiempo de espera del ticket, nivel de urgencia y condición de cliente premium.
- **Acciones (A):** mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3.
- **Entorno (E):** mesa de ayuda, clientes, técnicos de soporte, sistema de tickets y servicios tecnológicos.
- **Objetivo:** asignar cada ticket al nivel de soporte adecuado para resolverlo de manera rápida y eficiente.
- **Medida de desempeño (P):** tiempo promedio de atención, cantidad de tickets resueltos, satisfacción del cliente y cumplimiento de los tiempos establecidos.
