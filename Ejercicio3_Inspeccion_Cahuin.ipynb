## Ejercicio 3: Inspección visual de un producto

### Ficha PEAS

- **Percepción (S):** matriz de 5 × 5 píxeles capturada por una cámara, donde 0 representa un píxel normal y 1 representa un píxel defectuoso.
- **Acciones (A):** aprobar el producto, enviarlo a revisión manual o rechazarlo.
- **Entorno (E):** línea de producción, productos fabricados, cámara de inspección y área de control de calidad.
- **Objetivo:** detectar productos defectuosos y evitar que continúen en el proceso de producción o lleguen al cliente.
- **Medida de desempeño (P):** precisión en la detección de defectos, rapidez de inspección, reducción de productos defectuosos aprobados y bajo número de productos correctos rechazados.
